```python
balance = 1000

def deposit(current_balance, amount):
    current_balance = current_balance + amount
    print(f"{amount} deposited")
    return current_balance

def withdraw(current_balance, amount):
    if amount > current_balance:
        print(f"Cannot withdraw {amount}, insufficient balance")
    else:
        current_balance = current_balance - amount
        print(f"{amount} withdrawn")
    return current_balance

def check_balance(current_balance):
    print(f"Current balance: {current_balance}")


while True:
    print("\n1. Deposit")
    print("2. Withdraw")
    print("3. Check Balance")
    print("4. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        amount = int(input("Enter amount: "))
        balance = deposit(balance, amount)

    elif choice == "2":
        amount = int(input("Enter amount: "))
        balance = withdraw(balance, amount)

    elif choice == "3":
        check_balance(balance)

    elif choice == "4":
        print(f"Thank you")
        break

    else:
        print(f"Invalid choice")

```

```python
class Weather:

    def __init__(self, t1, t2, t3, h1, h2, h3):
        self.t1 = t1
        self.t2 = t2
        self.t3 = t3
        self.h1 = h1
        self.h2 = h2
        self.h3 = h3

    def avg_temperature(self):
        avg = (self.t1 + self.t2 + self.t3) / 3
        print(f"Average Temperature: {avg}")

        if avg > 30:
            print("Weather is Sunny")
        elif avg > 20:
            print("Weather is Normal")
        else:
            print("Weather is Cold")

    def avg_humidity(self):
        avg = (self.h1 + self.h2 + self.h3) / 3
        print(f"Average Humidity: {avg}")

t1 = float(input("Enter temp 1: "))
t2 = float(input("Enter temp 2: "))
t3 = float(input("Enter temp 3: "))

h1 = float(input("Enter humidity 1: "))
h2 = float(input("Enter humidity 2: "))
h3 = float(input("Enter humidity 3: "))

w = Weather(t1, t2, t3, h1, h2, h3)

w.avg_temperature()
w.avg_humidity()
```
